# Custom CNN from Scratch Result Analysis

This notebook analyzes the `Custom CNN` model trained from scratch for binary diabetic retinopathy detection.
The goal is to document how a non-pretrained baseline behaves before comparing it with transfer learning models.


```python
from pathlib import Path
import sys

import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import torch
from PIL import Image
from sklearn.metrics import accuracy_score, confusion_matrix, f1_score, precision_score, recall_score, roc_auc_score, roc_curve
from torch.utils.data import DataLoader

sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (8, 5)
```

```python
project_root = Path.cwd().resolve()
for candidate in [project_root, *project_root.parents]:
    if (candidate / "src").exists() and (candidate / "configs").exists():
        project_root = candidate
        break

sys.path.insert(0, str(project_root / "src"))

from dr_detection.config import load_yaml_config, resolve_path
from dr_detection.dataset import FundusImageDataset, load_fundus_records, default_transforms
from dr_detection.train import build_model

config = load_yaml_config(project_root / "configs" / "custom_cnn_scratch.yaml")
dataset_cfg = config["dataset"]
training_cfg = config["training"]

checkpoint_path = project_root / "models" / "custom_cnn_scratch" / "best_model.pt"
figures_dir = project_root / "outputs" / "custom_cnn"
figures_dir.mkdir(parents=True, exist_ok=True)

label_map = {0: "No DR", 1: "DR Present"}

test_records = load_fundus_records(
    csv_path=resolve_path(project_root, dataset_cfg["test_csv_path"]),
    images_dir=resolve_path(project_root, dataset_cfg["test_images_dir"]),
    image_extension=dataset_cfg.get("image_extension", ".png"),
    binary_classification=dataset_cfg.get("binary_classification", True),
)

test_dataset = FundusImageDataset(
    test_records,
    transform=default_transforms(
        image_size=dataset_cfg.get("image_size", 224),
        train=False,
        preprocessing_cfg=dataset_cfg.get("preprocessing"),
    ),
)
test_loader = DataLoader(
    test_dataset,
    batch_size=training_cfg.get("batch_size", 32),
    shuffle=False,
    num_workers=0,
)

if torch.cuda.is_available():
    device = torch.device("cuda")
elif torch.backends.mps.is_available():
    device = torch.device("mps")
else:
    device = torch.device("cpu")

checkpoint = torch.load(checkpoint_path, map_location=device)
model = build_model(
    model_name=checkpoint["model_name"],
    num_classes=checkpoint["num_classes"],
    pretrained=False,
)
model.load_state_dict(checkpoint["model_state_dict"])
model = model.to(device)
model.eval()

print(f"Checkpoint: {checkpoint_path}")
print(f"Device: {device}")
print(f"Test size: {len(test_dataset)}")
```

```python
all_labels = []
all_predictions = []
all_scores = []
all_image_paths = []

with torch.no_grad():
    for images, labels, image_paths in test_loader:
        images = images.to(device)
        logits = model(images)
        probabilities = torch.softmax(logits, dim=1)
        predictions = probabilities.argmax(dim=1)

        all_labels.extend(labels.tolist())
        all_predictions.extend(predictions.cpu().tolist())
        all_scores.extend(probabilities[:, 1].cpu().tolist())
        all_image_paths.extend(image_paths)

predictions_df = pd.DataFrame(
    {
        "image_path": all_image_paths,
        "true_label": all_labels,
        "predicted_label": all_predictions,
        "dr_probability": all_scores,
    }
)
predictions_df["true_class"] = predictions_df["true_label"].map(label_map)
predictions_df["predicted_class"] = predictions_df["predicted_label"].map(label_map)
predictions_df["correct"] = predictions_df["true_label"] == predictions_df["predicted_label"]

predictions_df.head()
```

**Table 5.1. Final test-set metrics for the custom CNN trained from scratch**


```python
metrics_df = pd.DataFrame(
    {
        "metric": ["accuracy", "precision", "recall", "f1_score", "roc_auc"],
        "value": [
            accuracy_score(all_labels, all_predictions),
            precision_score(all_labels, all_predictions, zero_division=0),
            recall_score(all_labels, all_predictions, zero_division=0),
            f1_score(all_labels, all_predictions, zero_division=0),
            roc_auc_score(all_labels, all_scores),
        ],
    }
)
metrics_df["value"] = metrics_df["value"].round(4)
metrics_df.to_csv(figures_dir / "table_5_1_custom_cnn_metrics.csv", index=False)
metrics_df
```

**Table 5.2. Confusion-matrix counts for the custom CNN on the test split**


```python
cm = confusion_matrix(all_labels, all_predictions)
tn, fp, fn, tp = cm.ravel()

cm_counts_df = pd.DataFrame(
    {
        "outcome": ["True Negative", "False Positive", "False Negative", "True Positive"],
        "count": [tn, fp, fn, tp],
    }
)
cm_counts_df.to_csv(figures_dir / "table_5_2_custom_cnn_confusion_counts.csv", index=False)
cm_counts_df
```

**Figure 5.1. Confusion matrix for the custom CNN trained from scratch**


```python
fig, ax = plt.subplots(figsize=(6, 5))
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    cbar=False,
    xticklabels=[label_map[0], label_map[1]],
    yticklabels=[label_map[0], label_map[1]],
    ax=ax,
)
ax.set_title("Confusion Matrix for the Custom CNN")
ax.set_xlabel("Predicted Label")
ax.set_ylabel("True Label")
plt.tight_layout()
fig.savefig(figures_dir / "figure_5_1_custom_cnn_confusion_matrix.png", dpi=300, bbox_inches="tight")
plt.show()
```

**Figure 5.2. ROC curve for the custom CNN trained from scratch**


```python
fpr, tpr, _ = roc_curve(all_labels, all_scores)
roc_auc = roc_auc_score(all_labels, all_scores)

fig, ax = plt.subplots(figsize=(6, 5))
ax.plot(fpr, tpr, linewidth=2, label=f"ROC curve (AUC = {roc_auc:.4f})")
ax.plot([0, 1], [0, 1], linestyle="--", color="gray", label="Random classifier")
ax.set_title("ROC Curve for the Custom CNN")
ax.set_xlabel("False Positive Rate")
ax.set_ylabel("True Positive Rate")
ax.legend(loc="lower right")
plt.tight_layout()
fig.savefig(figures_dir / "figure_5_2_custom_cnn_roc_curve.png", dpi=300, bbox_inches="tight")
plt.show()
```

**Image 5.1. Misclassified retinal fundus images for the custom CNN**


```python
misclassified_df = predictions_df.loc[~predictions_df["correct"]].copy().head(6)
if misclassified_df.empty:
    print("No misclassified images found.")
else:
    fig, axes = plt.subplots(2, 3, figsize=(12, 8))
    axes = axes.flatten()

    for ax, (_, row) in zip(axes, misclassified_df.iterrows()):
        image = Image.open(row["image_path"]).convert("RGB")
        ax.imshow(image)
        ax.set_title(
            f"true={row['true_class']}\npred={row['predicted_class']}\nP(DR)={row['dr_probability']:.3f}"
        )
        ax.axis("off")

    for ax in axes[len(misclassified_df):]:
        ax.axis("off")

    plt.tight_layout()
    fig.savefig(figures_dir / "image_5_1_custom_cnn_misclassified_examples.png", dpi=300, bbox_inches="tight")
    plt.show()
```
